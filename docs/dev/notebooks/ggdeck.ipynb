# `ggdeck` Feature

This notebook demonstrates the usage of the `ggdeck` function to combine several plots into one overlaid figure with multiple y-axes.

```python
import numpy as np
from lets_plot import *
LetsPlot.setup_html()
```

## Generate Mock Data

We create a dataset with multiple signals having different magnitudes to demonstrate how `ggdeck` can handle mixed scales.

```python
np.random.seed(42)
L = 100
x = np.arange(L)

def gen_data(multiplier, standard_dev):
    return np.random.normal(multiplier * 10, multiplier * standard_dev, L)

plots = []
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']

for i, color in enumerate(colors, start=1):
    y = gen_data(i, 5)
    data = {'x': x, 'y': y}
    if i % 2 == 0:
        p = ggplot(data, aes(x='x', y='y')) + geom_line(color=color, linetype='dashed' if i > 2 else 'solid', size=1)
    else:
        p = ggplot(data, aes(x='x', y='y')) + geom_point(color=color, size=2, tooltips=layer_tooltips().line('xy|@x,@y'))
    plots.append(p)
```

## Basic Usage

By default, `ggdeck` alternates axes on the left and right sides without needing to manually specify the sides. If `sides` is not provided, it defaults to `'L'` and then `'R'` for the rest. We provide two plots here. Notice that the axis colors are automatically styled to match the geometry.

```python
ggdeck([plots[0], plots[1]])
```

## Customizing Sides

You can specify the placement of axes for each plot using the `sides` argument (`'L'` for Left, `'R'` for Right). 
When multiple axes share a side, they are automatically shifted outwards to prevent overlapping.

```python
deck = ggdeck(
    plots,
    sides=['L', 'R', 'L', 'R']
)

deck + theme_bw()
```

## Disable Auto Styling

Through `auto_style=False`, the axes will fallback to default theme styling rather than dynamically extracting and matching the plot geometry's color, fill, and linetype.

```python
ggdeck([plots[0], plots[1]], sides=['L', 'R'], auto_style=False)
```

## Empty / Transparent Cells

If `None` is provided for a cell, no axes or plot area is drawn for that layer.

```python
ggdeck([plots[0], None, plots[2]], sides=['L', 'R', 'R'])
```
